## 2. Configure the source data and output path
Download `dataframe.csv` from the source repository and keep it outside version control.


```python
from pathlib import Path

DATA_PATH = Path('../data/raw/dataframe.csv')
OUTPUT_DIR = Path('../results/bogota_notebook')
```

## 3. Verify the dataset version


```python
from oculoplacenta_temporalformer.data import sha256_file, verify_bogota_checksum

print('SHA-256:', sha256_file(DATA_PATH))
print('Matches manuscript version:', verify_bogota_checksum(DATA_PATH, strict=False))
```

## 4. Run the complete benchmark


```python
from oculoplacenta_temporalformer.benchmark import run_benchmark

performance, audit = run_benchmark(
    data_path=DATA_PATH,
    output_dir=OUTPUT_DIR,
    seed=20260731,
    n_splits=5,
    n_repeats=2,
    n_bootstrap=500,
)
performance
```
